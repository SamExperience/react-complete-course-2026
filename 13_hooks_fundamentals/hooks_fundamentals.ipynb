## LESSON 36 — What a Hook is

`useState` is a Hook. You have used it for eleven lessons, and nothing has told you what
that word means.

### The definition

> *Hooks* let you use different React features from your components.

That is the whole idea. A component is a plain JavaScript function — it has no memory, no
connection to React, no way to reach anything outside itself. A Hook is the doorway. Calling
`useState` is how a function asks React for a piece of memory; every other Hook asks for some
other React feature.

And they are recognisable by name:

> Functions whose names start with `use` are called *Hooks* in React.

That convention is not decoration. It is how you, your editor and React's own lint rules tell
a Hook from an ordinary function — which matters, because Hooks come with rules that ordinary
functions do not have. Those rules are LESSON 37.

### The map

React groups its built-in Hooks into categories. Here is the whole set, with the honest
answer about where each one appears in this course:

| category | Hooks | in this course |
|---|---|---|
| **State** | `useState`, `useReducer` | LESSON 25 · LESSON 52-54 |
| **Context** | `useContext` | LESSON 56 |
| **Ref** | `useRef`, `useImperativeHandle` | LESSON 49-51 · not covered |
| **Effect** | `useEffect`, `useLayoutEffect`, `useInsertionEffect`, `useEffectEvent` | LESSON 39-44 · the rest not covered |
| **Performance** | `useMemo`, `useCallback`, `useTransition`, `useDeferredValue` | LESSON 70 · LESSON 72 · the rest not covered |
| **Other** | `useDebugValue`, `useId`, `useSyncExternalStore`, `useActionState` | `useActionState` at LESSON 81 · the rest not covered |

Two things are worth reading off that table.

**You do not need most of it.** Six Hooks carry almost all real React work — `useState`,
`useEffect`, `useRef`, `useContext`, `useReducer`, and one of the performance pair when you
have measured a reason. The rest exist for specific situations you will recognise when you
meet them, and React's own docs mark several as *rarely used*.

**The ones you know already have company.** `useReducer` is in the same category as
`useState` because it does the same job differently, and that is exactly how LESSON 52 will
introduce it.

### What each of the six does, in one line

**The React API.** You are not using these yet — this is orientation, so that the names stop
being noise when you meet them.

| | React's own description |
|---|---|
| `useState` | "declares a state variable that you can update directly" |
| `useReducer` | "declares a state variable with the update logic inside a reducer function" |
| `useEffect` | "connects a component to an external system" |
| `useRef` | "declares a ref. You can hold any value in it, but most often it's used to hold a DOM node" |
| `useContext` | "reads and subscribes to a context" |
| `useMemo` | "lets you cache the result of an expensive calculation" |

Read `useEffect`'s line twice. *Connects a component to an external system* — not "runs code
after render", which is how it is usually mis-taught and how you will mis-use it. Topic 14
starts from that sentence.

### You can write your own

> You can also define your own custom Hooks as JavaScript functions.

A custom Hook is a function whose name starts with `use` and which calls other Hooks. It is
not a new feature — it is a way of packaging the ones above so two components can share
behaviour rather than copying it. Topic 21 covers them, and by then you will have written the
duplication that makes them obviously worth it.

### Key Notes

- A Hook lets a component use a React feature. `useState` asks React for memory; the others
  ask for something else.
- Hooks are the functions whose names begin with `use` — the convention is how the rules find
  them.
- Six Hooks cover almost everything; the rest are situational and several are rarely used.
- A custom Hook is your own function that calls other Hooks (topic 21).

## LESSON 37 — The Rules of Hooks

LESSON 25 gave you a rule with no reason attached: keep `useState` at the top of the
component, never inside a function, condition or loop. Here is the reason, and it is worth
knowing rather than obeying.

### The two rules

> **Only call Hooks at the top level.** Don't call Hooks inside loops, conditions, nested
> functions, or `try`/`catch`/`finally` blocks. Instead, always use Hooks at the top level of
> your React function, before any early returns.

> **Only call Hooks from React functions.** Don't call Hooks from regular JavaScript
> functions.

React's list of what that rules out is worth having in full:

> - Do not call Hooks inside conditions or loops.
> - Do not call Hooks after a conditional `return` statement.
> - Do not call Hooks in event handlers.
> - Do not call Hooks in class components.
> - Do not call Hooks inside functions passed to `useMemo`, `useReducer`, or `useEffect`.
> - Do not call Hooks inside `try`/`catch`/`finally` blocks.

And who may call them:

> - Call them at the top level in the body of a function component.
> - Call them at the top level in the body of a custom Hook.

### Why — React identifies your Hooks by order, not by name

This is the sentence the whole lesson exists for.

When you write `const [name, setName] = useState("")`, React does not learn that you called it
`name`. It cannot: the name is a local variable in your function and React never sees it. What
React has is **position** — this was the first Hook call in this component, so it gets the
first slot of that component's memory. The second call gets the second slot.

So the order of your Hook calls *is* the identity of your state. Keep the order identical on
every render and everything lines up. Change it once and every Hook after the change is
reading somebody else's slot.

### Key Notes

- Hooks must be called at the **top level** of a component or custom Hook — never in a
  condition, loop, nested function, `try`/`catch`, or after an early `return`.
- React matches Hooks to their state by **call order**, not by name. That is the entire
  reason for the rule.
- A conditional Hook shifts every later Hook by one slot, which is silent corruption, not a
  crash — until React notices the count changed.
- `use` is not a Hook, and is not bound by the top-level rule.

### Example

**Runnable — plain JS.** **Conceptual model — not React's actual implementation.**

React stores a component's Hook state somewhere and walks it in order on every render. That
ordering is the part worth seeing; the rest of React's machinery is not modelled here and this
cell should not be read as how React works internally.

````typescript
// Conceptual model — not React's actual implementation.
function l37makeComponent() {
  const slots = [];
  let cursor = 0;

  return {
    startRender() { cursor = 0; },
    useStateModel(initial) {
      const index = cursor;
      cursor = cursor + 1;
      if (slots.length <= index) slots[index] = initial;   // first render only
      return [slots[index], (next) => { slots[index] = next; }];
    },
    slots,
  };
}

const l37c = l37makeComponent();

// --- render 1: two Hook calls, in this order ---
l37c.startRender();
let [l37name, l37setName] = l37c.useStateModel("Ada");
let [l37count, l37setCount] = l37c.useStateModel(0);
console.log("render 1:", l37name, l37count);

l37setName("Grace");
l37setCount(5);

// --- render 2: SAME order, so everything lines up ---
l37c.startRender();
[l37name] = l37c.useStateModel("Ada");
[l37count] = l37c.useStateModel(0);
console.log("render 2:", l37name, l37count, "<- correct");

// --- render 3: the first call is skipped, as a condition would skip it ---
l37c.startRender();
// if (somethingFalsy) { useStateModel("Ada") }   <- not called this time
[l37count] = l37c.useStateModel(0);
console.log("render 3:", l37count, "<- the count is now reading the NAME's slot");
console.log("slots are still:", JSON.stringify(l37c.slots));
````

Nothing threw. The data was not lost. The second Hook simply read the first Hook's slot,
because slots are handed out by counting, and this render counted differently.

That is why the rule is absolute rather than a style preference.

### What React actually says when you break it

React does notice, and the message depends on how you broke it. All three of these come from
React 19:

```text
Rendered more hooks than during the previous render.
```

```text
Rendered fewer hooks than expected. This may be caused by an accidental early return statement.
```

```text
Invalid hook call. Hooks can only be called inside of the body of a function component.
```

That last one lists three causes, and the middle one is the one to read first: *"You might be
breaking the Rules of Hooks."* The other two — mismatched React versions, or two copies of
React in one app — are real but rarer.

React also warns about the order changing, and prints a Previous render / Next render table
showing exactly which call moved.

> **A lint rule does this for you.** The `eslint-plugin-react-hooks` rules catch almost every
> violation before you run the code. The scaffold this course uses does not install it, so for
> now the rule lives in your head — but knowing *why* it exists is what makes the lint error
> readable when you eventually meet it.

### `use` is the thing that looks like an exception and is not

React 19 has an API called `use`, and it can be called conditionally. That sounds like it
breaks the first rule, so it is worth being precise about why it does not:

> Despite its name, `use` is not a Hook. Unlike Hooks, it can be called inside loops and
> conditional statements like `if`.

It is not a Hook that is exempt from the rules — it is not a Hook. The rules bind Hooks, and
`use` is a different kind of thing that happens to share the prefix. LESSON 77 covers what it
actually does; all you need today is that meeting it in code is not a counter-example to this
lesson.

## LESSON 38 — Render and commit

You have said "re-render" a hundred times in this course. This lesson makes it precise,
because the next two topics are hard to read without it: topic 14 places Effects relative to
these steps, and topic 23's whole argument is about which of them actually costs anything.

### Three steps, in order

React's own breakdown:

1. **Triggering** a render
2. **Rendering** the component
3. **Committing** to the DOM

They are separate, they happen in that order, and most confusion about React comes from
treating them as one event.

### 1. Trigger

A render starts for exactly one of two reasons:

- it is the component's **first** render, or
- **state changed** — *"Updating your component's state automatically queues a render."*

That is the whole list. Nothing else you do causes a render: not editing the DOM by hand, not
changing a plain variable, not time passing. LESSON 25's two reasons for state were the same
fact from the other side.

### 2. Render

> **"Rendering" is React calling your components.**

That is all the word means. React calls your function, your function returns JSX describing
what the UI should be, and React does the same for every component inside it. At the end React
has a complete description of what the screen *should* look like — and has not touched the
screen.

This is why a render can be a pure calculation. Your component is a function of its props and
state; running it should not change anything outside it.

### 3. Commit

Now React compares the description with what is already on the page and applies the
differences — and only the differences:

> **React only changes the DOM nodes if there's a difference between renders.**

Then the browser takes over:

> After rendering is done and React updated the DOM, the browser will repaint the screen.

### You have already measured this

LESSON 35 was this lesson without the vocabulary. Pressing `+1` in experiment 15 produced:

```text
   render — CountDisplay, count = 1
   render — ShoppingList
   [profiler] id=counter phase=update
   [profiler] id=sidebar phase=update

DOM changes in the counter subtree:  1
DOM changes in the list subtree:     0
```

Both components were **rendered** — React called both functions. Only one subtree was
**committed** to, because only one produced a difference. The shopping list's HTML was
byte-for-byte identical before and after, so there was nothing to apply.

Say that back in the new words and the whole of LESSON 35 collapses into one sentence:
*rendering is not committing, and committing nothing is the normal case.*

### What this buys you

| when someone says | they mean |
|---|---|
| "it re-rendered" | React called the component function again |
| "it didn't update" | either no render was triggered, or the render produced no difference |
| "the DOM changed" | the commit step had something to apply |

Those are three different problems with three different causes, and until this lesson you had
one word for all of them. When something does not appear on screen, the useful first question
is now **which step didn't happen** — was no render triggered (no setter ran, or the value was
`Object.is`-equal), or did the render produce the same output it did last time?

### Key Notes

- Three steps, in order: **trigger → render → commit**, then the browser paints.
- A render is triggered by the first mount or by a **state change**, and by nothing else.
- *"Rendering is React calling your components"* — it produces a description, and touches
  nothing.
- Committing applies only the differences, so a render that changes nothing commits nothing.
