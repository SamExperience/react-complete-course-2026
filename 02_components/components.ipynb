## LESSON 4 — A component is a function

A **component** is a JavaScript function that returns JSX, the HTML-looking syntax you saw
in `App.jsx`.

**The React API:**

```jsx
function Greeting() {
  return <p>Hello from a component.</p>;
}
```

You then use it as if it were a tag:

```jsx
<Greeting />
<Greeting />
```

Written once, used as often as you like. That is the whole reason components exist.

Topic 3 is entirely about JSX and its rules. For now you need exactly three facts:

1. the JSX goes in the `return`;
2. a component returns **one** element — if you need two side by side, wrap them in a
   container for now, and LESSON 10 will show you a tidier wrapper;
3. a component is exported and imported like any other function.

### The capital letter is not a style choice

JSX decides what `<Greeting />` means by looking at the **first letter**.

- **`<Greeting />`** — capital. React calls your function.
- **`<greeting />`** — lowercase. React treats it as an HTML tag and creates an empty
  `<greeting></greeting>` element. Your function is never called.

Nothing crashes, which is what makes it confusing. You get a blank space and this in the
console:

```text
The tag <greeting> is unrecognized in this browser.
If you meant to render a React component, start its name with an uppercase letter.
```

The playground experiment has a commented-out `<greeting />` at the bottom. Uncomment it,
look at the page and the console, then comment it back — it logs that error on every
reload, and you want a clean console for the next two lessons.

### Key Notes

- A component is a function that returns JSX, named with a **capital letter**.
- Use it as a tag: `<Greeting />`.
- One component returns **one** element — wrap siblings in a container for now.
- A lowercase name is read as an HTML tag, and your function never runs.

### Example

**The React API.** A component, and a second component using it twice. `export default`
is what lets another file import it.

```jsx
function Greeting() {
  return <p>Hello from a component.</p>;
}

export default function Experiment() {
  return (
    <div>
      <h1>Two greetings</h1>
      <Greeting />
      <Greeting />
    </div>
  );
}
```

## LESSON 5 — Composition with `children`

Some components are **containers**. A card, a panel, a page section, a dialog — their job
is to wrap something, and they have no business knowing what that something is.

React passes every component exactly one argument: an object. This lesson needs one single
thing out of it, called **`children`** — whatever was nested between the opening and
closing tags.

**The React API:**

```jsx
function Card({ children }) {
  return <section>{children}</section>;
}
```

```jsx
<Card>
  <h2>Revenue</h2>
  <p>12,400 EUR</p>
</Card>
```

The `{children}` inside `Card` means *put whatever I was given right here*. `Card`
contributes the `<section>` wrapper; the caller contributes the contents.

That object React passes is called **props**, and topic 6 is entirely about it. `children`
is the one piece you need today — take it as given for now.

### Why this is worth the trouble

Compare two ways of building five dashboard cards:

- **With `children`** — one `Card` component, used five times with different contents.
- **Without** — `RevenueCard`, `OrdersCard`, `RefundsCard`, `VisitorsCard`,
  `ConversionCard`, each containing its own hardcoded text and its own copy of the
  wrapper. Change the border and you change it five times.

A container that does not know what it contains gets reused. One that does, does not.

A self-closing `<Card />` simply has no children, so `{children}` renders nothing. That is
not an error.

### Key Notes

- React gives every component one argument, an object; `children` is whatever was nested
  inside the tag.
- `{children}` is where that content gets placed.
- A container component should not know what it contains.
- The rest of the props system is topic 6.

### Example

**The React API.** One wrapper, two different uses. `Card` is written once.

```jsx
function Card({ children }) {
  return (
    <section>
      <hr />
      {children}
      <hr />
    </section>
  );
}

export default function Dashboard() {
  return (
    <div>
      <Card>
        <h2>Revenue</h2>
        <p>12,400 EUR</p>
      </Card>

      <Card>
        <p>A second card, with completely different contents.</p>
      </Card>
    </div>
  );
}
```

## LESSON 6 — Deciding where components begin and end

Nothing forces you to split a UI up. One enormous component works, right until you cannot
find anything in it. So the real question is *when*.

### Four signals that a chunk wants to be a component

1. **You can name it.** If "that's the header", "that's the activity list" comes out of
   your mouth naturally, you have already found the boundary. Struggling for a name is a
   sign there isn't one.
2. **It has one job.** A component you can describe without saying "and" is the right size.
3. **You are scrolling** inside a single function to find the bit you want to edit.
4. **It repeats.** Three copies of the same markup is a nudge — though not an order.

### Signals you are splitting too early

- The name is `Wrapper2`, `Section3` or `Content`. No name, no boundary.
- The chunk is two lines, used once, and now lives in another file you have to open.
- You are splitting because you read that components should be small.

**Start coarse and split when it hurts.** Splitting later is easy. Merging three
components back into one, after you built them wrong, is not.

### What you can extract today

You have `children`, and nothing else yet. So an extracted component can:

- carry its **own fixed content** — `<Header />` always renders the same thing; or
- be a **wrapper** that takes `children`.

What it cannot do yet is show *different data each time it is used*. That needs props,
which is topic 6 — the very next one. You will bump into this limit in the first
mini-project, and that is on purpose: the fix means much more once you have felt the
problem.

### Key Notes

- If you can give the chunk a good name, it probably wants to be a component.
- One component, one job — describable without "and".
- Repetition is a signal, not an order. Three clear copies beat one confusing abstraction.
- Until topic 6, an extracted component either carries fixed content or takes `children`.

### Example

**The React API.** The same page, before and after. Note that the split version reads like
a table of contents.

```jsx
// Before — everything in one place
export default function App() {
  return (
    <div>
      <header>
        <h1>Acme</h1>
        <nav>…</nav>
      </header>
      <section>
        <h2>Revenue</h2>
        <p>12,400 EUR</p>
      </section>
      <footer>
        <p>© Acme</p>
      </footer>
    </div>
  );
}
```

```jsx
// After — three names you can read at a glance
export default function App() {
  return (
    <div>
      <SiteHeader />
      <RevenuePanel />
      <SiteFooter />
    </div>
  );
}
```
