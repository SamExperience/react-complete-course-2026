## LESSON 1 — UI as a function of state

In the JavaScript course you changed the page by **commanding** it. Something happened, so
you found an element and told it what to do:

**Not runnable here** — this is the imperative DOM code from the JavaScript course,
shown for contrast. There is no `document` in this kernel.

```js
const badge = document.querySelector("#cart-badge");
if (items === 0) {
  badge.textContent = "Your cart is empty";
  badge.classList.add("empty");
} else {
  badge.textContent = `${items} items`;
  badge.classList.remove("empty");
}
```

That is **imperative**: you write the exact steps that get the page from how it looks now
to how it should look. You are giving turn-by-turn directions.

React goes the other way. You write a function that answers one question — **what should
this look like, given this data?** — and React works out which parts of the page to change.
That is **declarative**: you name the destination, not the turns.

```js
// The same idea, declared instead of commanded.
function cartLabel(state) {
  return state.items === 0 ? "Your cart is empty" : `${state.items} items`;
}
```

The second version never touches the page. It takes data in and returns a description out.
Give it the same data twice and you get the same answer twice — which is exactly what lets
React re-run it whenever the data changes.

Neither model is wrong. Commanding the DOM directly is fine for one small change. The
declarative model earns its keep when many parts of the screen depend on the same data and
you are tired of keeping them in sync by hand.

One thing about the shape above: `cartLabel` returns a **string**, because a string is all
a notebook can show you. A real React component returns **JSX** — a description of UI
rather than a line of text — and topic 2 starts exactly there.

### Key Notes

- **Imperative:** you write the steps that change the page. **Declarative:** you describe
  the result and React applies it.
- A React UI is written as a **function of state**: the same state in gives the same UI out.
- React's own docs call the different looks a component can have its **visual states**.
- Once you use React, `document.querySelector` disappears from your day-to-day
  code. There is one narrow exception, in topic 16, and that is all.

### Example

**Runnable — plain JS.** No React here, just the shape of the idea: one function, state in,
description out. Run it and notice that the function is the only thing that decides what
each state looks like.

````typescript
function l1RenderCart(state) {
  if (state.items === 0) return "Your cart is empty";
  return `${state.items} item(s) — total ${state.total} EUR`;
}

// Same function, different state, different UI.
console.log(l1RenderCart({ items: 0, total: 0 }));
console.log(l1RenderCart({ items: 3, total: 47.5 }));

// Same state twice — same answer twice.
const l1State = { items: 3, total: 47.5 };
console.log(l1RenderCart(l1State) === l1RenderCart(l1State));
````

## LESSON 2 — Creating a React project with Vite

React is a library, not a program you can run. Browsers do not understand JSX, so
something has to transform your files before they reach the page. **Vite** does that in two
different ways: while you work it serves each module on demand, transformed as the browser
asks for it; when you are finished it produces a bundled production build.

One command sets everything up:

```bash
npm create vite@latest my-app -- --template react
cd my-app
npm install
npm run dev
```

`--template react` means **React with JavaScript**. There is a TypeScript variant too;
this course does not use it. Passing the template as a flag also means the command asks
you nothing — run `npm create vite@latest` on its own and it becomes interactive instead.

### What you get

| file or folder | what it is for |
|---|---|
| `index.html` | the only real HTML page. It holds `<div id="root">` and loads `/src/main.jsx` |
| `src/main.jsx` | the entry point: finds `#root` and renders your app into it |
| `src/App.jsx` | your first component — the top of your UI |
| `src/index.css`, `src/App.css` | styles (topic 4) |
| `public/` | files copied to the site root untouched (`favicon.svg`, `icons.svg`) |
| `src/assets/` | images you `import` from code; the build processes and renames them |
| `vite.config.js` | Vite's configuration — this is where the React plugin is registered |
| `package.json` | dependencies, and the scripts you run |
| `.oxlintrc.json` | config for `oxlint`, the linter the template ships with. Optional; nothing here depends on it |

### The scripts

| command | what it does |
|---|---|
| `npm run dev` | starts the dev server, reloads the page as you save |
| `npm run build` | produces the production build in `dist/` |
| `npm run preview` | serves the built `dist/` locally so you can check it |
| `npm run lint` | runs `oxlint` — optional |

Open `src/App.jsx` and you will see HTML-looking tags sitting inside a JavaScript file.
That is **JSX**, and topic 3 is entirely about it. For now, leave it alone.

### Key Notes

- `npm create vite@latest . -- --template react` scaffolds React + JavaScript and asks

- `index.html` is the only page in the app. Everything else is JavaScript that fills

- `dev` while you work, `build` for production, `preview` to check what `build` produced.
- `public/` is copied as-is; `src/assets/` is processed by the build.

## LESSON 3 — How the app starts

A React app has one way in, and it is worth being able to trace it from memory. Four steps:

1. **`index.html`** holds an empty `<div id="root"></div>` and loads one script:
   `/src/main.jsx`.
2. **`src/main.jsx`** finds that div and hands it to React.
3. **`src/App.jsx`** exports the component at the top of your UI.
4. From then on, **React owns everything inside `#root`**.

**The React API** — this is the real `src/main.jsx` that Vite generates:

```jsx
import { StrictMode } from 'react'
import { createRoot } from 'react-dom/client'
import './index.css'
import App from './App.jsx'

createRoot(document.getElementById('root')).render(
  <StrictMode>
    <App />
  </StrictMode>,
)
```

Read it as two separate jobs. `createRoot(domNode)` claims a DOM node for React and gives
you back a *root*. `root.render(...)` puts your app inside it — and the first time it runs,
React clears any HTML that was already in that node. An app built entirely in React
normally has exactly one `createRoot` call.

`document.getElementById('root')` is ordinary DOM code, and it is the last you will see of
it. Get the id wrong and it returns `null`, so React throws **"Target container is not a
DOM element"** — a message worth recognising, because the cause is almost always a typo or
a script that ran before the div existed.

### What StrictMode is

`<StrictMode>` is a wrapper that turns on extra checks **in development only**. With it on,
React renders your components an extra time and runs each Effect through an extra
setup-and-cleanup cycle, to expose bugs caused by impure code or missing cleanup. None of
this happens in a production build.

Mentioning it now saves confusion later: when something logs twice and you have only one
component, this is why. Topic 14 returns to it where it actually matters.

`import './index.css'` is not something browsers can do — importing CSS from JavaScript is
a build-tool feature. Topic 4 covers it.

### Key Notes

- The chain is `index.html` → `main.jsx` → `createRoot(...).render(<App />)` → `App`.
- `createRoot` needs a real DOM node. A wrong id gives **"Target container is not a DOM
  element"**.
- React owns everything inside the root node, and clears it on the first render.
- `StrictMode` adds development-only checks — it changes nothing in production.

### Example

**The React API.** The whole chain, as three files. Nothing else is involved in getting
your first component on screen.

```html
<!-- index.html -->
<body>
  <div id="root"></div>
  <script type="module" src="/src/main.jsx"></script>
</body>
```

```jsx
// src/main.jsx
import { createRoot } from 'react-dom/client'
import App from './App.jsx'

createRoot(document.getElementById('root')).render(<App />)
```

```jsx
// src/App.jsx
export default function App() {
  return <h1>Hello</h1>
}
```
