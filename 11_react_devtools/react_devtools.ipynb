## LESSON 34 — The Components tab

You have already used browser DevTools in this course — LESSON 12 had you inspect a box and
work out which CSS rule won. That was the **Elements** panel, and it showed you the DOM.

React ships its own panel, and it shows something the DOM cannot.

### Installing it

React Developer Tools is a browser extension. From react.dev:

> Use React Developer Tools to inspect React components, edit props and state, and identify
> performance problems.

Install it from **react.dev/learn/react-developer-tools**, which links to the official builds
for **Chrome**, **Firefox** and **Edge**. Then:

> Now, if you visit a website **built with React**, you will see the *Components* and
> *Profiler* panels.

Two panels. This lesson is entirely about **Components**; Profiler is LESSON 35.

### Two trees, and they are not the same tree

Open the playground experiment and look at both panels side by side.

- **Elements** shows the **DOM**: `<div>`, `<section>`, `<ul>`, `<li>` — what the browser
  actually has.
- **Components** shows **your components**: `Experiment14`, `Panel`, `SearchBox`,
  `MemberList`, `MemberCard`.

They are not two views of one thing. Measured on experiment 14:

```text
components in the React tree:   8
DOM elements inside #root:     17
```

A component is not an element. `Panel` produces a `<section>` and an `<h2>` and then whatever
it was given as `children`. `MemberList` produces a `<ul>`. A component can produce many
elements, one, or none at all — and the DOM keeps no record of which component produced what.
That record is exactly what the Components tab has and the Elements panel does not.

### What a selected component tells you

Select a component in the tree and you get the two things this course has spent nine topics
on:

| | what it is | where it came from |
|---|---|---|
| **props** | what this component was handed | its parent |
| **state** | what this component owns | its own `useState` calls, in the order you wrote them |

Here is the real tree of experiment 14, with the values the panel shows:

```text
App
└─ Experiment14                    state: "" , "m2"
   └─ Panel                        props: title="Team", children=…
      ├─ SearchBox                 props: value="", onChange=ƒ
      └─ MemberList                props: members=[3 items], leadId="m2"
         ├─ MemberCard             props: member={Ada Lovelace…}, isLead=false
         ├─ MemberCard             props: member={Grace Hopper…}, isLead=true
         └─ MemberCard             props: member={Katherine Johnson…}, isLead=false
```

Notice `children` showing up as a prop on `Panel`. LESSON 5 told you `children` was an
ordinary prop; here is the tool agreeing.

### Why this beats the Elements panel for React questions

The Elements panel can tell you **what is on the screen**. It cannot tell you **why**.

Look at the tree above and at the page. One row has a star next to it. In the DOM that star is
a `<span>` — true, and useless. In the Components tab it is `isLead={true}` on a particular
`MemberCard`, which came from `leadId` on `MemberList`, which came from state in
`Experiment14`. The tool does not just show you the value; it shows you **the chain the value
travelled down**, which is where the answer to "why does it look like this?" always lives.

That chain is the real skill. Reading a wrong value on screen means walking **up** the tree
until you reach the component that produced it.

### Locating the component you care about

Two directions, and you will use both:

- **Downwards** — start at the root and walk the tree. Good when you know roughly where you
  are going, and the fastest way to learn an unfamiliar codebase's shape.
- **Upwards** — start from the thing you can see on the page and find the component behind it.
  Every DevTools build has a way to pick an element on the page and jump to it; find it once
  and you will use it constantly.

The tree also shrinks and grows as the app runs. Type in the search box in experiment 14 and
watch `MemberCard` entries disappear — that is `.map()` from LESSON 19 rendering fewer items,
and the tree is the most direct view of it you will ever get.

> **Editing values.** React's docs describe DevTools as a way to *edit* props and state, not
> only read them — handy for checking "what would this look like if the flag were true?"
> without touching the code. The exact affordance depends on your version of the extension, so
> find it yourself rather than trusting a screenshot. Anything you change this way is gone on
> the next reload: it changes the running app, never your source.

### Key Notes

- The **Components** tab shows your component tree; the **Elements** panel shows the DOM.
  Different trees, different sizes, different questions.
- Selecting a component shows its **props** (from its parent) and its **state** (its own).
- Use it to answer *why does the UI look like this* by following a value up the tree to where
  it is owned.
- It is an extension you install once, from the links on react.dev.

## LESSON 35 — The Profiler: measuring before deciding

LESSON 34's Components tab answers *what is the tree right now*. This lesson answers a
different question: *what happened when the user did that?*

### The problem it solves

Sooner or later something feels slow, or you suspect a component is doing far more work than
it should. At that point there are two ways to proceed:

1. guess which component is at fault and start changing code, or
2. find out.

The second one is the job of the Profiler, and the whole lesson is really one sentence:
**first find out what rendered, then decide whether anything needs doing.** Deciding what to
do about it is topic 23, much later. Right now you are learning to look.

### Two different things share the name "Profiler"

This trips people up, so get it straight before anything else:

| | what it is | how you get it |
|---|---|---|
| **`<Profiler>`** | a real React component you `import` and put in your JSX | it is part of React |
| **the Profiler panel** | a tab in React Developer Tools, next to Components | the browser extension from LESSON 34 |

They measure the same underlying thing. One you write into your code and read in the console;
the other is a UI you record an interaction with. This lesson uses the **API** for the part
you can prove to yourself, and points you at the **panel** for the part that is genuinely a
tool.

### The `<Profiler>` API

```jsx
import { Profiler } from "react";

<Profiler id="sidebar" onRender={onRender}>
  <ShoppingList items={items} />
</Profiler>
```

> `<Profiler>` lets you measure rendering performance of a React tree programmatically.

`id` is a string naming the part of the UI you are measuring. `onRender` is called every time
the tree inside it commits, with six arguments:

```js
function onRender(id, phase, actualDuration, baseDuration, startTime, commitTime) { }
```

| | |
|---|---|
| `id` | the `id` of the `<Profiler>` that just committed |
| `phase` | `"mount"`, `"update"` or `"nested-update"` |
| `actualDuration` | ms spent rendering this tree for this update |
| `baseDuration` | ms estimating a full re-render of the subtree without optimisations |
| `startTime` / `commitTime` | timestamps for when React began and finished |

You can use several `<Profiler>`s to measure different parts of an app, and you can nest them.

> **Do not learn the numbers.** Durations differ on every machine and on every run — a value
> measured here was `0.39…` ms and the next one was different. What is stable, and what you
> should read, is **`id` and `phase`**: which part committed, and whether it was a first mount
> or an update. Treat durations as something to *compare within one session*, never as a fact
> about the app.

> Profiling is not free, so: *"it is disabled in the production build by default."*

### What the measurement actually shows

Experiment 15 has two separately measured subtrees and one button. Press `+1` and this is the
real output:

```text
   render — CountDisplay, count = 1
   render — ShoppingList
   [profiler] id=counter phase=update
   [profiler] id=sidebar phase=update
```

Both subtrees rendered. Both committed. Now the part that matters:

```text
DOM changes in the counter subtree:  1
DOM changes in the list subtree:     0
```

The shopping list's HTML was **byte-for-byte identical** before and after. `ShoppingList` ran
again, produced the same elements, React compared them with what was already there, and
changed nothing.

### So a re-render is not a bug

That is the most important idea in this lesson. A component rendering again is React doing
its ordinary job: your component is a function, something above it changed, so React called it
to find out what the UI should be now. Most of the time the answer is "the same as before" and
React touches nothing.

Seeing a component in a profile therefore tells you **it ran** — not that it was wrong, not
that it was slow, and not that it needs fixing. It is a fact to explain, not a fault to
correct.

Something is only worth acting on when it is both **expensive** and **avoidable**, and you
cannot know either without measuring first. That is why this lesson comes before the tools
that change things: `memo`, `useMemo`, `useCallback` and React Compiler are all topic 23, and
every one of them is a bad idea applied to a problem you have not confirmed you have.

### Components tab or Profiler panel?

| you want to know | use |
|---|---|
| what is in the tree, and what props/state does it have *right now* | **Components** (LESSON 34) |
| what rendered while the user did something, and how the work was spread | **Profiler** |

And neither is the browser's own Performance or Elements panel, which know about frames,
scripts and DOM nodes but nothing about your components.

### Key Notes

- `<Profiler>` is a React API; the Profiler **panel** is part of React DevTools. Same idea,
  two different things.
- `onRender` reports `id` and `phase` (`"mount"` / `"update"` / `"nested-update"`) plus
  timings — read the first two, and never memorise the timings.
- Rendering again is normal, and it often changes nothing in the DOM. Measured: two subtrees
  re-rendered, one made zero DOM changes.
- Measure first. What to *do* about a real problem is topic 23.
