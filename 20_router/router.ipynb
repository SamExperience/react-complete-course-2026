## LESSON 58 — Why client-side routing

### What a browser does on its own

Click an `<a href="/about">` and the browser does something drastic: it throws away the current
page entirely, asks the server for a new document, and starts again from nothing. New HTML, new
JavaScript, new everything.

For a site of documents that is exactly right. For an application it is a disaster, because
**everything in memory is gone**:

- the state in every component — LESSON 25's whole subject
- the data you fetched — topic 15's whole subject
- the scroll position, the open panel, the half-typed search

The user sees a white flash and your app reloads itself to show a different screen.

### What a router does instead

A **client-side router** intercepts the navigation. Instead of letting the browser fetch a new
document, it:

1. changes the URL in the address bar,
2. works out which of your components matches that URL,
3. renders it.

No request, no reload, no lost state. The URL still means something — it can be bookmarked,
shared, and the Back button still works — but the page never goes away.

That combination is the whole point: **the addressability of a website with the continuity of an
application.**

### The package, and a version note worth reading

```bash
npm i react-router
```

The package is **`react-router`**, currently **version 8**, and everything is imported from it:

```jsx
import { BrowserRouter, Routes, Route, Link, NavLink } from "react-router";
```

> **If you search the web you will find `react-router-dom` everywhere.** That was the browser
> package for years, and an enormous amount of existing code and tutorials import from it. It
> stopped at version 7; version 8 lives in `react-router`. If an example you find imports from
> `react-router-dom`, it is not wrong — it is older. Use `react-router` in new work, and expect
> to meet the other name constantly.

This is the third time the course has met this shape: `forwardRef` (LESSON 49) and
`<Context.Provider>` (LESSON 56) are the same deal — one form to write, one form to recognise.

### Setting it up

**The React API.** Wrap your app once, at the root:

```jsx
import { BrowserRouter, Routes, Route } from "react-router";

ReactDOM.createRoot(root).render(
  <BrowserRouter>
    <Routes>
      <Route path="/" element={<App />} />
    </Routes>
  </BrowserRouter>,
);
```

Three pieces, each with one job:

| | |
|---|---|
| `<BrowserRouter>` | wraps the app once and connects it to the browser's URL |
| `<Routes>` | picks **one** matching `<Route>` and renders it |
| `<Route path element>` | says which component belongs to which URL |

`<Routes>` choosing exactly one match is worth noting — it is not a list of everything that
could render, it is a decision.

### Navigating: `Link`, not `<a>`

This is the single most important line in the lesson:

```jsx
import { Link } from "react-router";

<Link to="/concerts/salt-lake-city">Concerts</Link>
```

An `<a href>` tells the **browser** to navigate, which reloads everything. A `<Link to>` tells
the **router** to navigate, which does not. They look identical on screen — `Link` renders a
real `<a>` underneath, so it is still a proper link you can middle-click, copy, or open in a new
tab — but a normal click is intercepted.

Use `<a href>` only for leaving your app: an external site, a download, a `mailto:`.

### `NavLink` for navigation that knows where it is

> `NavLink` makes it easy to show active states

```jsx
<NavLink to="/" className={({ isActive }) => (isActive ? "active" : "")}>
  Home
</NavLink>
```

`NavLink` is a `Link` that tells you whether it points at the current URL. `className` — and
`style`, and even `children` — can be a **function** receiving `{ isActive }`, which is how a
navigation bar highlights the page you are on without any state of your own.

Notice what that replaces: without it you would keep the "current page" in state, set it on
every click, and keep it in step with the URL by hand. The router already knows; asking it is
better than duplicating it (LESSON 29, in yet another setting).

### Key Notes

- A client-side router changes the URL and swaps components **without reloading**, so state
  and fetched data survive.
- The package is **`react-router`** (v8); imports come from `"react-router"`. `react-router-dom`
  is the older name you will see online.
- `<BrowserRouter>` wraps once · `<Routes>` picks one match · `<Route path element>` maps URL
  to component.
- Navigate with `<Link to>`, never `<a href>`, inside your app. `NavLink` adds `isActive`.

## LESSON 59 — URL params and nested routes

Two routes is a demo. Real apps have URLs that carry data and layouts that survive navigation.

### Dynamic segments

A path segment beginning with `:` becomes a **dynamic segment** — a slot that matches anything
and hands you what it matched:

```jsx
<Route path="teams/:teamId" element={<Team />} />
```

> If a path segment starts with `:` then it becomes a "dynamic segment". When the route matches
> the URL, the dynamic segment will be parsed from the URL and provided as `params` to other
> router APIs like `useParams`.

```jsx
import { useParams } from "react-router";

export default function Team() {
  let params = useParams();
  // params.teamId
}
```

The name after the colon is the key you read back. `/teams/17` gives `params.teamId === "17"`.

**It is always a string.** URLs have no types, so `"17"` is what you get — and `params.teamId
=== 17` is `false`. If you need a number, convert it deliberately, exactly as you did for form
inputs in LESSON 33.

### The URL as state you did not have to write

This is worth pausing on, because it changes how you design a screen.

In Mini-project 3 the selected employee lived in `useState`. That works, and it has a
limitation you may not have noticed: **the selection cannot be shared, bookmarked, or restored
by the Back button**, because it exists only in memory.

Move it into the URL — `/employees/3` — and all three come free. The router is now holding a
piece of your application's state, and it is the one piece the browser already knows how to
persist, share and navigate.

That is the real reason to reach for a param, and it is a design decision rather than a
technical one: **is this something a user might want to link to?** If yes, it belongs in the
URL.

### Nested routes and `<Outlet>`

> Routes can be nested inside parent routes.

```jsx
<Routes>
  <Route path="dashboard" element={<Dashboard />}>
    <Route index element={<Home />} />
    <Route path="settings" element={<Settings />} />
  </Route>
</Routes>
```

> The path of the parent is automatically included in the child, so this config creates both
> `"/dashboard"` and `"/dashboard/settings"` URLs. Child routes are rendered through the
> `<Outlet/>` in the parent route.

```jsx
import { Outlet } from "react-router";

export default function Dashboard() {
  return (
    <div>
      <h1>Dashboard</h1>
      {/* will either be <Home/> or <Settings/> */}
      <Outlet />
    </div>
  );
}
```

`<Outlet />` is a hole in the parent that the matched child fills. Everything around it —
the heading, a sidebar, a nav bar — stays mounted while the child changes, which means **its
state survives** navigation between children.

If that sounds familiar, it should: it is `children` from LESSON 5, with the router choosing
what goes in the slot instead of you.

### Index routes

> Index routes render into their parent's `<Outlet/>` at their parent's URL (like a default
> child route). They are configured with the `index` prop.

```jsx
<Route path="dashboard" element={<Dashboard />}>
  <Route index element={<DashboardHome />} />
  <Route path="settings" element={<Settings />} />
</Route>
```

Without the index route, `/dashboard` renders `Dashboard` with an **empty** outlet — a heading
and nothing else. The index route is the answer to "what should show when the user is at the
parent's own URL?", and forgetting it is the commonest nested-routing bug.

### Key Notes

- `:param` in a path captures a segment; read it with `useParams()`, and it is always a
  **string**.
- Put in the URL whatever a user might want to link to — the URL is state the browser persists
  for you.
- Nested routes render into the parent's `<Outlet />`; everything around it stays mounted and
  keeps its state.
- An **index route** is what renders at the parent's own URL. Without one the outlet is empty.

## LESSON 60 — Navigating in code, and the route that catches everything

`Link` covers almost everything. This lesson is the rest.

### `useNavigate`, and when not to use it

> This hook allows the programmer to navigate the user to a new page without the user
> interacting.

```jsx
import { useNavigate } from "react-router";

export function LoginPage() {
  let navigate = useNavigate();

  return (
    <MyLoginForm
      onSuccess={() => {
        navigate("/dashboard");
      }}
    />
  );
}
```

Read the warning that comes with it **before** the API, because this is a Hook people reach for
far too early:

> For normal navigation, it's best to use `Link` or `NavLink`. They provide a better default
> user experience like keyboard events, accessibility labeling, "open in new window", right
> click context menus, etc.
>
> Reserve usage of `useNavigate` to situations where the user is *not* interacting but you need
> to navigate, for example:
> - After a form submission completes
> - Logging them out after inactivity
> - Timed UIs like quizzes, etc.

Every item on that list has the same shape: **nothing was clicked.** A form finished, a timer
expired, a session ended. When the user *does* click something that takes them elsewhere, that
is a link, and a real `<a>` underneath buys you middle-click, right-click, keyboard access and
screen-reader labelling for free.

A button with an `onClick` that calls `navigate` throws all of that away to look like a link.

### The common shape: navigate after a successful action

This is topic 10's form with one line added:

```jsx
function handleSubmit(event) {
  event.preventDefault();

  const errors = validate(values);
  setErrors(errors);
  if (Object.keys(errors).length > 0) return;

  addEmployee(values);
  navigate("/employees");     // only after success
}
```

Note where the call is: **after** the validation guard, so a failed submit stays on the page
with the user's input intact (LESSON 32's rule). Navigating away from a form that failed is one
of the more infuriating bugs you can ship.

### The catch-all route

Nothing you have written so far handles a URL that matches no route. The user gets a blank
page — no error, no message, nothing to click.

React Router's catch-all uses `*`:

> Also known as "catchall" and "star" segments. If a route path pattern ends with `/*` then it
> will match any characters following the `/`, including other `/` characters.

Used at the top level, a bare `*` matches anything that got this far:

```jsx
<Routes>
  <Route path="/" element={<Home />} />
  <Route path="teams/:teamId" element={<Team />} />
  <Route path="*" element={<NotFound />} />
</Routes>
```

`<Routes>` picks **one** match, and the catch-all is the one that matches when nothing better
did. A 404 page in your app is not a nicety — without it a typo in a shared link produces a
white screen, which looks exactly like a broken app.

A useful `NotFound` says what happened, and gives a way out:

```jsx
function NotFound() {
  return (
    <>
      <h1>That page does not exist</h1>
      <Link to="/">Back to the directory</Link>
    </>
  );
}
```

### Two different "not found"s

Worth separating, because they are handled in different places:

| | what it means | where it is handled |
|---|---|---|
| `/nonsense` | **no route matches** | the catch-all `*` route |
| `/teams/999` | a route matched, **no such team** | inside `Team`, after the lookup fails |

The second is LESSON 46's empty state wearing a URL, and the router cannot help with it — it
matched the route correctly. Only your component knows that team 999 does not exist.

### Key Notes

- `useNavigate` is for navigation the user did **not** click: after a submit, a timeout, a
  logout. Use `Link` for everything else.
- Navigate **after** the success path, never before the validation guard.
- `<Route path="*">` catches URLs nothing else matched. Without it a bad link renders nothing.
- "No route matched" and "the route matched but the thing does not exist" are different
  problems in different places.

## LESSON 61 — `useSearchParams` for filters and pagination

LESSON 59 put one value in the URL — an id, as part of the path. This lesson puts the *rest* of
a screen's state there: which filter, which page, what was searched.

### First, the platform thing

The query string is the part after `?`:

```text
/employees?q=ada&department=research&page=2
```

The browser gives you an object for reading and building these, and it is plain JavaScript with
no React involved:

```js
const params = new URLSearchParams("q=ada&page=2");
params.get("q");            // "ada"
params.get("missing");      // null
params.set("page", "3");
params.toString();          // "q=ada&page=3"
```

Three things to know before you use it, all of which the example cell demonstrates:

- **Everything is a string.** `params.get("page")` gives `"2"`, never `2`.
- **A missing key is `null`**, not `undefined` — so `??` and a default work nicely together.
- **It encodes for you.** `set("q", "grace hopper")` produces `q=grace+hopper`, and reading it
  back gives the space again. Never build a query string by hand with `+` and `&`; you will
  get it wrong for spaces, ampersands and accented characters.

### The Hook

> Returns a tuple of the current URL's `URLSearchParams` and a function to update them. Setting
> the search params causes a navigation.

```jsx
import { useSearchParams } from "react-router";

const [searchParams, setSearchParams] = useSearchParams();

const query = searchParams.get("q") ?? "";
const page = Number(searchParams.get("page") ?? "1");
```

It is a pair, like `useState` — and the comparison is exact enough to be the point of the
lesson. The difference is **where the value lives**: not in memory, but in the URL, which means
it is shared, bookmarked and navigated by the Back button for free.

Setting accepts several shapes:

```jsx
setSearchParams({ tab: "1" });                       // an object
setSearchParams("?tab=1");                           // a string
setSearchParams((params) => {                        // a callback, to change one key
  params.set("page", "2");
  return params;
});
```

The callback form is the one you will use most, because it keeps the other parameters — exactly
like the updater function in LESSON 26, and for the same reason.

### The caveat that will catch you

> The `searchParams` object is mutable, but if you change it without calling `setSearchParams`,
> its values will change between renders if some other state causes the component to re-render
> and the URL will not reflect the values.

So `searchParams.set("page", "2")` on its own does **something** — it edits the object — and
nothing you wanted: the URL does not change and the screen does not update. It is LESSON 27's
mutation bug in a new costume, and the cure is the same: go through the setter.

### What belongs in the URL, and what does not

Not everything. The test is the one from LESSON 59, sharpened:

**If a user opened this exact link tomorrow, should they see what you are seeing?**

| | |
|---|---|
| a search term, a filter, a page number, a sort order | **URL** — all of it is "what am I looking at" |
| whether a dropdown is open, a half-typed value, a hover | **state** — nobody wants to link to that |

There is one more consideration, and it is the reason this is not a free win: **setting search
params causes a navigation.** Every change is a history entry, so a search box that writes to
the URL on every keystroke makes the Back button useless — thirty presses to escape one word.
Debounce it (LESSON 42), or write on submit rather than on change.

### Key Notes

- `URLSearchParams` is a browser API: everything is a **string**, a missing key is **`null`**,
  and it handles encoding.
- `useSearchParams()` returns `[searchParams, setSearchParams]` — a pair like `useState`, with
  the value living in the URL.
- Use the **callback form** to change one key without losing the others.
- Never mutate `searchParams` directly; the URL will not change.
- Put in the URL what someone might share a link to. Remember each change is a history entry.

### Example

**Runnable — plain JS.** `URLSearchParams` is the browser's, not React's, so it runs here
perfectly well — and getting comfortable with it away from a component is worth five minutes.

````typescript
// Reading
const l61params = new URLSearchParams("q=ada&department=research&page=2");

console.log("q          ->", l61params.get("q"));
console.log("page       ->", l61params.get("page"), "| typeof:", typeof l61params.get("page"));
console.log("missing    ->", l61params.get("sort"));
console.log("with default ->", l61params.get("sort") ?? "name");

// Writing — note the encoding
const l61next = new URLSearchParams(l61params);
l61next.set("q", "grace hopper & co");
l61next.set("page", "1");
console.log("");
console.log("encoded    ->", l61next.toString());
console.log("read back  ->", l61next.get("q"), "<- the space and & survive the round trip");

// The original is untouched, because we copied it first
console.log("original   ->", l61params.toString());

// A helper of the shape you will write in a component
function l61withParam(params, key, value) {
  const next = new URLSearchParams(params);
  if (value === "" || value === null) {
    next.delete(key);
  } else {
    next.set(key, value);
  }
  return next;
}

console.log("");
console.log("set sort   ->", l61withParam(l61params, "sort", "name").toString());
console.log("clear q    ->", l61withParam(l61params, "q", "").toString());
````

Note `l61withParam` copies before changing. That is the same instinct as LESSON 27 — and here
it is not optional, because the object you were handed is the one the router is holding.
